```python
# ================================
# 🌳 Decision Tree - PrimeVul (JSON FIXED)
# ================================

import os
import json
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report

# ================================
# 🔹 1. LOAD DATA CORRECTLY (JSON)
# ================================
data_path = "/kaggle/input/datasets/nikunjnawal009/primevul-dt"
file_path = os.path.join(data_path, os.listdir(data_path)[0])

data = []

with open(file_path, "r") as f:
    for line in f:
        try:
            data.append(json.loads(line))
        except:
            continue

df = pd.DataFrame(data)

print("Columns:", df.columns)
print("Shape:", df.shape)

# ================================
# 🔹 2. SELECT CORRECT COLUMNS
# ================================
X = df["func"]      # code
y = df["target"]    # label

# Clean
X = X.fillna("").astype(str)
y = y.fillna(0).astype(int)

# ================================
# 🔹 3. TRAIN-TEST SPLIT
# ================================
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.1, random_state=42
)

# ================================
# 🔹 4. TF-IDF
# ================================
vectorizer = TfidfVectorizer(max_features=5000, ngram_range=(1,2))

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

# ================================
# 🔹 5. DECISION TREE
# ================================
model = DecisionTreeClassifier(max_depth=20, random_state=42)
model.fit(X_train_tfidf, y_train)

# ================================
# 🔹 6. PREDICT
# ================================
y_pred = model.predict(X_test_tfidf)

# ================================
# 🔹 7. RESULTS
# ================================
print("\n📊 RESULTS")
print("Accuracy :", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall   :", recall_score(y_test, y_pred))
print("F1 Score :", f1_score(y_test, y_pred))

print("\n🔍 Classification Report:\n")
print(classification_report(y_test, y_pred))
```
